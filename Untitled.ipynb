```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow  # Import cv2_imshow for Colab compatibility

# Function to load an image
def load_image(path):
    image = cv2.imread(path)
    if image is None:
        print("Error: Could not load image")
        exit()
    return image

# Function to translate the image
def translate_image(image, tx, ty):
    # Get the image dimensions (height, width)
    height, width = image.shape[:2]

    # Define the translation matrix (2D affine transformation)
    translation_matrix = np.float32([[1, 0, tx], [0, 1, ty]])

    # Apply the translation
    translated_image = cv2.warpAffine(image, translation_matrix, (width, height))
    return translated_image

# Function to rotate the image
def rotate_image(image, angle, scale=1.0):
    # Get the image dimensions (height, width)
    height, width = image.shape[:2]

    # Get the rotation matrix for the given angle and scale
    center = (width // 2, height // 2)
    rotation_matrix = cv2.getRotationMatrix2D(center, angle, scale)

    # Apply the rotation
    rotated_image = cv2.warpAffine(image, rotation_matrix, (width, height))
    return rotated_image

# Function to scale the image
def scale_image(image, scale_x, scale_y):
    # Get the image dimensions (height, width)
    height, width = image.shape[:2]

    # Calculate the new dimensions after scaling
    new_width = int(width * scale_x)
    new_height = int(height * scale_y)

    # Resize the image
    scaled_image = cv2.resize(image, (new_width, new_height))
    return scaled_image

# Function to resize images to the same dimensions
def resize_to_original_size(images):
    # Resize all images to match the size of the first image (original image)
    height, width = images[0].shape[:2]
    resized_images = [cv2.resize(img, (width, height)) for img in images]
    return resized_images

# Combine transformations into one image
def combine_transformations(image, tx, ty, angle, scale_rotation, scale_x, scale_y):
    # Apply translation, rotation, and scaling
    translated_image = translate_image(image, tx, ty)
    rotated_image = rotate_image(image, angle, scale_rotation)
    scaled_image = scale_image(image, scale_x, scale_y)

    # Resize all transformed images to match the original image size
    resized_images = resize_to_original_size([image, translated_image, rotated_image, scaled_image])

    # Stack all transformed images side by side (horizontally)
    combined_image = np.hstack(resized_images)
    return combined_image

# Main function to demonstrate loading, translating, rotating, and scaling the image
def main():
    # Load the image
    image_path = input("Enter the path of the image file: ")
    image = load_image(image_path)

    # Get transformation parameters
    tx = int(input("Enter translation in X direction (pixels): "))
    ty = int(input("Enter translation in Y direction (pixels): "))
    angle = float(input("Enter the rotation angle (in degrees): "))
    scale_rotation = float(input("Enter the scaling factor for rotation (default is 1.0): "))
    scale_x = float(input("Enter the scaling factor in X direction (e.g., 1.2 for 120%): "))
    scale_y = float(input("Enter the scaling factor in Y direction (e.g., 1.2 for 120%): "))

    # Combine all transformations
    combined_image = combine_transformations(image, tx, ty, angle, scale_rotation, scale_x, scale_y)

    # Display the combined image with all transformations
    cv2_imshow(combined_image)  # Display the combined image

    # Automatically save the combined image
    cv2.imwrite('combined_transformed_image.jpg', combined_image)
    print("Combined image saved as 'combined_transformed_image.jpg'")

if __name__ == "__main__":
    main()
```
