```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread(r'/content/dish.jpg')

# Convert from BGR (OpenCV default) to RGB for matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Get image dimensions
X, Y, D = img_rgb.shape
div1 = X // 2
div2 = Y // 2

# Define the four quadrants
top_left = img_rgb[:div1, :div2]
top_right = img_rgb[:div1, div2:]
bottom_left = img_rgb[div1:, :div2]
bottom_right = img_rgb[div1:, div2:]

# Prepare the figure for displaying
fig, axs = plt.subplots(2, 2, figsize=(10, 10))

# Display the four quadrants with scaling
axs[0, 0].imshow(top_left)
axs[0, 0].set_title(f"Top Left\n({div1} x {div2})")
axs[0, 0].axis('on')  # Keep axes on to show scale
axs[0, 0].set_aspect('auto')  # Auto scale the aspect ratio

axs[0, 1].imshow(top_right)
axs[0, 1].set_title(f"Top Right\n({div1} x {div2})")
axs[0, 1].axis('on')  # Keep axes on to show scale
axs[0, 1].set_aspect('auto')  # Auto scale the aspect ratio

axs[1, 0].imshow(bottom_left)
axs[1, 0].set_title(f"Bottom Left\n({div1} x {div2})")
axs[1, 0].axis('on')  # Keep axes on to show scale
axs[1, 0].set_aspect('auto')  # Auto scale the aspect ratio

axs[1, 1].imshow(bottom_right)
axs[1, 1].set_title(f"Bottom Right\n({div1} x {div2})")
axs[1, 1].axis('on')  # Keep axes on to show scale
axs[1, 1].set_aspect('auto')  # Auto scale the aspect ratio

# Adjust layout to avoid overlap and add more space at the bottom
plt.subplots_adjust(wspace=0.11, hspace=0.3, bottom=0.1)

# Show the plot
plt.show()
```
